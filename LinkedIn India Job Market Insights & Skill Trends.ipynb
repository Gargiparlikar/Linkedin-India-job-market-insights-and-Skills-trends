```python
%%writefile App.py
import streamlit as st
import pandas as pd
import plotly.express as px

# ================= PAGE CONFIG =================
st.set_page_config(
    page_title="India Job Market Skill Intelligence",
    page_icon="🇮🇳",
    layout="wide"
)

st.title("📊LinkedIn India Job Market Insights & Skill Trends")
st.markdown(
    "Analysis of LinkedIn India job postings focusing on "
    "**skills, industries, job levels, companies, and locations**."
)

# ================= LOAD DATA =================
@st.cache_data
def load_data():
    return pd.read_csv("final_data.csv")

df = load_data()

# ================= DATA CLEANING =================


# ================= INDUSTRY CLEANING =================
df["Industry"] = df["Industry"].astype(str).str.strip().str.lower()

industry_map = {
    # Technology / IT
    "information technology": "Technology, Information And Internet",
    "information technology & services": "Technology, Information And Internet",
    "internet": "Technology, Information And Internet",
    "computer software": "Technology, Information And Internet",
    "software development": "Technology, Information And Internet",

    # IT Consulting
    "it services": "It Services And It Consulting",
    "it services and consulting": "It Services And It Consulting",
    "information technology services": "It Services And It Consulting",

    # Telecom
    "telecommunications": "Telecommunications",
    "telecom": "Telecommunications",

    # Real Estate
    "real estate": "Real Estate",
    "commercial real estate": "Real Estate",
    "property": "Real Estate",

    # Recruitment / HR
    "staffing and recruiting": "Recruitment",
    "human resources": "Recruitment",

    # Finance
    "banking": "Finance",
    "financial services": "Finance",
    "investment banking": "Finance",
}

df["Industry"] = df["Industry"].replace(industry_map)

# Clean invalid values
df["Industry"] = df["Industry"].str.title()
df = df[~df["Industry"].isin(["Nan", "None", "", "Other"])]
df = df[df["Industry"].notna()]


# -------- JOB LEVEL CLEANING --------
df["Level"] = df["Level"].astype(str).str.strip().str.lower()

level_map = {
    "entry level": "Entry Level",
    "junior": "Entry Level",
    "associate": "Entry Level",
    "mid level": "Mid Level",
    "senior level": "Senior Level",
    "senior": "Senior Level",
    "lead": "Senior Level",
    "manager": "Manager",
    "director": "Director",
    "executive": "Executive"
}

df["Level"] = df["Level"].replace(level_map)
df["Level"] = df["Level"].str.title()
df = df[~df["Level"].isin(["Nan", "None", ""])]
df = df[df["Level"].notna()]

# ================= EXPERIENCE LEVEL DERIVATION =================

df["Level"] = df["Level"].astype(str).str.lower().str.strip()

df["Involvement"] = (
    df["Involvement"]
    .astype(str)
    .str.strip()
    .str.title()
)

# Optional: Rename to a cleaner column
df["Experience_Level"] = df["Involvement"]


# ================= SKILL COLUMNS =================
skill_cols = [
    "PYTHON","C++","JAVA","HADOOP","SCALA","FLASK","PANDAS","SPARK",
    "NUMPY","PHP","SQL","MYSQL","CSS","MONGODB","NLTK","TENSORFLOW",
    "LINUX","RUBY","JAVASCRIPT","DJANGO","REACT","REACTJS","AI","UI",
    "TABLEAU","NODEJS","EXCEL","POWER BI","SELENIUM","HTML","ML"
]

# ================= SIDEBAR FILTERS =================
st.sidebar.header("🔎 Filters")
st.sidebar.info("Filters are synced with cleaned data")

location_filter = st.sidebar.multiselect(
    "📍 Location",
    options=sorted(df["Location"].dropna().unique())
)

industry_filter = st.sidebar.multiselect(
    "🏭 Industry",
    options=sorted(df["Industry"].unique())
)

level_filter = st.sidebar.multiselect(
    "🎯 Job Level",
    options=sorted(df["Level"].unique())
)

experience_filter = st.sidebar.multiselect(
    "🧠 Experience Level",
    options=sorted(df["Experience_Level"].unique())
)





# ================= APPLY FILTERS =================
filtered = df.copy()

if location_filter:
    filtered = filtered[filtered["Location"].isin(location_filter)]
if industry_filter:
    filtered = filtered[filtered["Industry"].isin(industry_filter)]
if level_filter:
    filtered = filtered[filtered["Level"].isin(level_filter)]
if experience_filter:
    filtered = filtered[filtered["Experience_Level"].isin(experience_filter)]
 

# ================= KPI SECTION =================
st.divider()
c1, c2, c3, c4 = st.columns(4)

c1.metric("📄 Total Jobs", len(filtered))
c2.metric("🏢 Companies", filtered["Company_Name"].nunique())
c3.metric("📍 Locations", filtered["Location"].nunique())
c4.metric("👥 Avg Applicants",
          int(filtered["Total_applicants"].mean()) if not filtered.empty else 0)
top_exp = filtered["Experience_Level"].mode()[0] if not filtered.empty else "N/A"
c5 = st.columns(1)[0]
c5.metric("🎯 Most Hiring Experience Level", top_exp)

st.divider()

# ================= TOP SKILLS =================
st.subheader("🔥 Most In-Demand Skills")

if filtered.empty:
    st.warning("No data available for selected filters.")
else:
    skills = filtered[skill_cols].sum().sort_values(ascending=False).reset_index()
    skills.columns = ["Skill", "Demand"]

    fig = px.bar(
        skills.head(15),
        x="Demand",
        y="Skill",
        orientation="h",
        color="Demand",
        title="Top Skills Required by Employers in India"
    )
    st.plotly_chart(fig, use_container_width=True)

# ================= JOB LEVEL DISTRIBUTION =================
st.subheader("🎯 Job Level Distribution")

if not filtered.empty:
    lvl = filtered["Level"].value_counts().reset_index()
    lvl.columns = ["Level", "Jobs"]

    fig = px.pie(
        lvl,
        names="Level",
        values="Jobs",
        hole=0.4,
        title="Job Level Distribution"
    )
    st.plotly_chart(fig, use_container_width=True)

# ================= INDUSTRY DISTRIBUTION =================
st.subheader("🏭 Top Hiring Industries")

if not filtered.empty:
    ind = filtered["Industry"].value_counts().head(10).reset_index()
    ind.columns = ["Industry", "Jobs"]

    fig = px.bar(
        ind,
        x="Jobs",
        y="Industry",
        orientation="h",
        color="Jobs",
        title="Top Industries Hiring in India"
    )
    st.plotly_chart(fig, use_container_width=True)

# ================= LOCATION DISTRIBUTION =================
st.subheader("🌍 Top Hiring Locations")

if not filtered.empty:
    loc = filtered["Location"].value_counts().head(10).reset_index()
    loc.columns = ["Location", "Jobs"]

    fig = px.bar(
        loc,
        x="Jobs",
        y="Location",
        orientation="h",
        color="Jobs",
        title="Top Job Locations in India"
    )
    st.plotly_chart(fig, use_container_width=True)

# ================= HEATMAP =================
st.subheader("📊 Skill Demand Across Locations")

if not filtered.empty:
    top_locs = filtered["Location"].value_counts().head(8).index
    heatmap_df = (
        filtered[filtered["Location"].isin(top_locs)]
        .groupby("Location")[skill_cols]
        .sum()
    )

    if heatmap_df.sum().sum() > 0:
        fig = px.imshow(
            heatmap_df.T,
            aspect="auto",
            color_continuous_scale="Plasma",
            title="Skill Demand Heatmap (India)"
        )
        st.plotly_chart(fig, use_container_width=True)

 # ================= Experience Level Distribution ================= 
st.subheader("🧠 Experience Level Distribution")

exp_counts = (
    filtered["Experience_Level"]
    .value_counts()
    .reset_index()
)

exp_counts.columns = ["Experience Level", "Jobs"]

fig_exp = px.bar(
    exp_counts,
    x="Jobs",
    y="Experience Level",
    orientation="h",
    color="Jobs",
    title="Job Distribution by Experience Level"
)

st.plotly_chart(fig_exp, use_container_width=True)

# ================= Skills by Experience Level (Recruiter Favorite) ================= 
st.subheader("🔥 Skill Demand by Experience Level")

exp_skill_df = (
    filtered.groupby("Experience_Level")[skill_cols]
    .sum()
)

fig_exp_skill = px.imshow(
    exp_skill_df.T,
    aspect="auto",
    color_continuous_scale="Plasma",
    title="Skills Required Across Experience Levels"
)

st.plotly_chart(fig_exp_skill, use_container_width=True)

# ================= DOWNLOAD =================
st.subheader("💾 Download Filtered Data")

@st.cache_data
def to_csv(df):
    return df.to_csv(index=False).encode("utf-8")

st.download_button(
    "📥 Download CSV",
    to_csv(filtered),
    "filtered_india_jobs.csv",
    "text/csv"
)

# ================= RAW DATA =================
with st.expander("📄 View Raw Data"):
    st.dataframe(filtered, use_container_width=True)

st.divider()
st.caption("📌 Built by Gargi Parlikar | India Job Market Skill Intelligence Dashboard")
```
